### Other assignment operators

| Symbol | Operation | Example | Alternate Representation |
|--------|-----------|---------|--------------------------|
| += | Adds right operand to left operand and stores result on left side operand | `a += b` | `a = a + b` |
| -= | Subtracts right operand to left operand and stores result on left side operand | `a -= b` | `a = a - b` |
| *= | Multiplies right operand to left operand and stores result on left side operand | `a *= b` | `a = a * b` |
| /= | Divides right operand from left operand and stores result on left side operand | `a /= b` | `a = a / b` |
| %= | Divides right operand from left operand and stores the remainder of the result on left side operand | `a %= b` | `a = a % b` |
| //= | Divides right operand to left operand and stores the whole part of the quotient obtained on left side operand | `a //= b` | `a = a // b` |

````python
ball = 40
ball += 1

print(ball)
````

### Relational operators

- Tests numerical equalities/inequalities between different values and returns a boolean

- All operators have same precedence

---

| Symbol | Operation |
|--------|-----------|
| < | strictly less than |
| <= | less than or equal to |
| > | strictly greater than |
| >= | greater than or equal to |
| == | equal to |
| != | not equal to |

### Equal to(==):

````python
#Comparing batsmen scores
batsman1_score = 85
batsman2_score = 89

print(batsman1_score == batsman2_score)
````

### Not Equal to(!=):

````python
#Checking the teams scores
team1_score = 230
team2_score = 245

print(team1_score != team2_score)
````

### Greater than or Less than (>/<):

````python
# Comparing the runs scored by two teams
team1_score = 280
team2_score = 245

print(team1_score >= team2_score)
````

### Greater than or equal to(>=)

````python
#Comparing bowling averages of two bowlers
bowler1_average = 24.50
bowler2_average = 26.75

print(bowler1_average <= bowler2_average)
````

* Relational operators are useful in conditional statements to evaluate whether the condition holds good

### Logical operators

values

### AND Operator(and):

The AND operator returns True only if both the conditions are True.

````python
# Checking if a batsman scored a century and the team won the match
batsman_score = 120
is_team_won = True

if batsman_score >= 100 and is_team_won:
    print("The batsman scored a century and the team won the match.")
else:
    print("Either the batsman didn't score a century or the team lost the match.")
````

### OR Operator(or):

* The OR operator returns True if one of the condition is True

````python
# Checking if a team won the match or the match ended in a tie
is_team_won = True
is_match_tied = False

if is_team_won or is_match_tied:
    print("The team won the match or the match ended in a tie.")
else:
    print("The team lost the match.")
````

### NOT Operator(not):

* The NOT operator is used to negate a boolean value

````python
#Checking if a batsman is not out
is_out = False

if not is_out:
    print("The batsman is not out.")
else:
    print("The batsman is out.")
````

## 8 .Conditional Statements:<a class="anchor" id="conds"></a>

### if Statement:<a class="anchor" id="if"></a>

Syntax:<br>
        &emsp;if condition:<br>
            &emsp;&emsp;#body of if statement

If the condition is true, the body of if statement is executed.

````python
runs = 1500
wickets = 70
if runs>1000 and wickets > 100:
    print("Good all-rounder")
````

### if-else statement:<a class="anchor" id="ifelse"></a>

Syntax:<br>
        &emsp;if condition:<br>
            &emsp;&emsp;#body of if statement<br>
        &emsp;else:<br>
            &emsp;&emsp;#body of else statement    

If the condition is true, the body of if statement is executed. If the condition is false, then body of else statement is executed

````python
runs = 1500
wickets = 70
if runs>1000 and wickets > 100:
    print("Good all-rounder")
else:
    print("Part time bowler")
````
