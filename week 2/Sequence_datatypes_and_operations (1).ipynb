### String Slicing
- **:** is used to slice some elements from an ordered sequence
  - Syntax: **`object[start:stop:step]`**
  - The default value for start and step are **0** and **1** respectively
  - Negative value for step can be used to access the elements from the end of the sequence


````python
samplestr = 'Cricket'   # getting substring 'Cric' from 'Cricket'
samplestr[slice(4)]
````

````python
samplestr[slice(1,6,2)]  # getting substring 'rce'
````

````python
str4 = 'Bowler'

print(str4[0])
print(str4[:4])
print(str4[::-1])
````

## List

- A sequence data type which can contain multiple items of different data types
- A list can be created using a square bracket ```[]``` or by using the ```list()``` method
- In general, lists are ordered, mutable, and can take duplicate values

Creation:

````python
#list of numbers
score = [23, 50, 100, 100]
print("score:",score)
````

````python
teams = ['Australia','India','England']     #list of characters

mixed_list = [1, 2, 'India', 2.5,'IPL']        #list of mixed value types

print("Teams: ", teams, type(teams))
print("Mixed List: ", mixed_list, type(mixed_list))
````

````python
#Empty List
runs =[]
print("Runs:",runs, type(runs))
````

#### List Operations

````python
print(dir(teams))
````

````python
print(help(teams.append))
````

````python
new_team = "Pakistan"
teams.append(new_team) # Appends 'new_name' to the end of the list 'name_list'
print(teams)
````

````python
print(help(teams.reverse))
````

````python
teams.reverse() # Reverses the contents of the list, in-place
print(teams)
````

````python
print(len(teams)) # Prints the length (number of elements) of the list
````

### List Indexing and Slicing

````python
print(teams)
````

## Tuples

- A sequence data type which can contain multiple items of different data types
- A tuple can be created using parantheses ```()``` or by using the ```tuple()``` method
- In general, tuples are ordered, immutable, and can take duplicate values

Creation:

````python
player = ('Virat Kohli', 32, 'India')



print(player, type(player))
print(len(player))
````

````python
team = ('India', ['Virat Kohli', 'Rohit Sharma', 'Jasprit Bumrah'])
print(team, type(team))
print(len(team))
````

### Tuple Indexing and Slicing

````python
scores = (78, 102, 45, 91, 56)

print(scores[0])
print(scores[-1])
````

* Tuple Slicing

````python
scores = (78, 102, 45, 91, 56)

print(scores[1:4])
print(scores[:3])
print(scores[2:])
print(scores[::2])
````

* Unpacking Tuples - You can assign the elements of a tuple to multiple variables in a single assignment statement

````python
player = ('Virat Kohli', 32, 'India')
name, age, country = player

print(name)
print(age)
print(country)
````

### Tuple Concatenation

````python
tup = ('1, 2, 3')
print(tup, id(tup))
tup += (', 4')
print(tup, id(tup))
````

- This proves that tuples are immutable

### Range:
* A range is a sequence of numbers that can be created using the `range()` function

* Commonly used in loops and is especially helpful for iterating through a range of numbers.

* Function can take one, two, or three arguments, and it returns a sequence of numbers starting from a specified start value, up to (but not including) an end value, and with a specified step value.

    * range(stop)

    * range(start, stop)
    
    * range(start, stop, step)


Creation:

````python
overs = range(20)
for i in overs:
  print(i)

print("POWERPLAY Overs")
powerply_overs = range(1,7)
for i in powerply_overs:
  print(i)
````

````python
odd_runs = range(1,12,2)
print(odd_runs)
for i in odd_runs:
    print(i)
````

#### enumerate()

* Used to iterate through elements in an iterable (like a list, tuple, or string) while keeping track of the index (position) of each element in the iterable.
* returns an enumerate object, which is an iterator that generates pairs of index and value for each element in the iterable.

syntax:

  enumerate(iterable, start=0)


````python
# Enumerating a list with index and value pairs
players = ['Virat Kohli', 'Rohit Sharma', 'Jasprit Bumrah']

for index, players in enumerate(players, start = 1):
    print(f"Index {index}: {players}")
````
