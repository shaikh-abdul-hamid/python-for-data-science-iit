### Multiple Columns - Aggregation
Can apply multiple aggregation functions to the columns in each group by passing a dictionary of functions to apply to each column.

```python
# Group by player and calculate total runs, number of 100s, and number of 50s
player_stats = cricket_data.groupby('100').agg({
    'Runs': 'sum',
    '50': 'sum',
})

print(player_stats)
```

Typically used after a groupby operation to convert the grouped columns into regular columns in the DataFrame.
(hierarchical index->Flat Index)

```python
# Group by player and calculate total runs, number of 100s, and number of 50s
player_stats = cricket_data.groupby('100').agg({
    'Runs': 'sum',
    '50': 'sum',
}).reset_index()

print(player_stats)
```

### Multi-Indexed dataframes

```python
# Group by player and calculate total runs and batting average
player_stats = cricket_data.groupby(['100','50']).agg({
    'Runs': 'sum',
    '6s': 'sum',
})#.reset_index()
# Set player name as the index
#player_stats.set_index('100', inplace=True)

print(player_stats)
```

### Set Index

```python
# Group by player and calculate total runs and batting average
player_stats = cricket_data.groupby(['100','50']).agg({
    'Runs': 'sum',
    '6s': 'sum',
}).reset_index()
# Set index to half-century
player_stats.set_index('50', inplace=True)

print(player_stats)
```

## **Merging and Concatenating DataFrames**<a name="merge"></a>

### concat()
* The `concat()` function is used to concatenate multiple DataFrames.

```python
# Create sample DataFrames
df1 = pd.DataFrame({'Player': ['SuryaKumar Yadav', 'Rohit Sharma', 'Jasprit Bumrah'],
                    'Team': ['MI', 'MI', 'MI'],
                    'Runs': [11867, 9115,0 ],
                   'Wickets': [0,0,97]})

df2 = pd.DataFrame({'Player': ['Virat Kohli', 'Faf Duplesis', 'Maxwell'],
                    'Team': ['RCB', 'RCB', 'RCB'],
                    'Runs': [10115, 9115, 10767]})

df3 = pd.DataFrame({'Player': ['Ashwin', 'Siraj', 'Varun Chakravarthy'],
                    'Team': ['RR', 'RCB', 'England'],
                    'Wickets': [101, 91, 107]})

# Concat the DataFrames
concated_df = pd.concat([df1,df2,df3])

print(concated_df)
```

### merge()
* The `merge()` function is used to combine two or more DataFrames based on a common column

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Ruturaj Gaikwad'],
                    'Team': ['CSK', 'CSK', 'CSK'],
                    'Runs': [11867, 9115, 27]})

df2 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Kohli'],
                    'Team': ['CSK', 'CSK', 'RCB'],
                    'Average': [59.33, 48.55, 54.17]})

# Merge the DataFrames
merged_df = pd.merge(df1, df2, on='Player')

print(merged_df)
```

### Inner join()
* An inner join combines only the matching rows from two DataFrames based on common column.

```python
# Create two sample DataFrames
data1 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Ruturaj Gaikwad'],
                    'Team': ['CSK', 'CSK', 'CSK'],
                    'Runs': [11867, 9115, 27]})


data2 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Kohli'],
                    'Team': ['CSK', 'CSK', 'RCB'],
                    'Average': [59.33, 48.55, 54.17]})

# Perform inner join on the DataFrames
inner_join_df = pd.merge(data1, data2, on='Player', how='inner')

print(inner_join_df)
```

### Outer join()
#### Returns all the rows from both dataframes, and fills the missing values with NaN for any non-matching rows.

```python
# Create two sample DataFrames
df1 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Ruturaj Gaikwad'],
                    'Team': ['CSK', 'CSK', 'CSK'],
                    'Runs': [11867, 9115, 27]})
df2 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Kohli'],
                    'Team': ['CSK', 'CSK', 'RCB'],
                    'Average': [59.33, 48.55, 54.17]})

print(df1)
print("\n")
print(df2)
print("\n")
# Perform outer join on the DataFrames
outer_join_df = pd.merge(df1, df2, on='Player', how='outer')

print(outer_join_df)
```

### Left join()
#### Returns all the rows from the left dataframe and the matching rows from the right dataframe. If there is no matching row in the right dataframe, the missing values are filled with NaN.

```python
# Create two sample DataFrames
df1 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Ruturaj Gaikwad'],
                    'Team': ['CSK', 'CSK', 'CSK'],
                    'Runs': [11867, 9115, 27]})
df2 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Kohli'],
                    'Team': ['CSK', 'CSK', 'RCB'],
                    'Average': [59.33, 48.55, 54.17]})
print(df1)
print("\n")
print(df2)
print("\n")
# Perform left join on the DataFrames
left_join_df = pd.merge(df1, df2, on='Player', how='left')

print(left_join_df)
```

### Right join()
#### Returns all the rows from the right dataframe and the matching rows from the left dataframe. If there is no matching row in the left dataframe, the missing values are filled with NaN.

```python
# Create two sample DataFrames
df1 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Ruturaj Gaikwad'],
                    'Team': ['CSK', 'CSK', 'CSK'],
                    'Runs': [11867, 9115, 27]})
df2 = pd.DataFrame({'Player': ['MS Dhoni', 'Rahane', 'Kohli'],
                    'Team': ['CSK', 'CSK', 'RCB'],
                    'Average': [59.33, 48.55, 54.17]})

print(df1)
print("\n")
print(df2)
print("\n")
# Perform right join on the DataFrames
right_join_df = pd.merge(df1, df2, on='Player', how='right')

print(right_join_df)
```

#### 1. Find the players who have the maximum and minimum batting average

```python
# Find the player who has maximum and minimum batting average
max_average = cricket_data['Bat_Avg'].idxmax()

min_average = cricket_data['Bat_Avg'].idxmin()

print(f"Player with maximum batting average:{cricket_data.loc[max_average,'Player']}")
print(f"Player with minimum batting average:{cricket_data.loc[min_average,'Player']}")
```

#### 2. Find the players who has maximum sixes and best strike rate.

```python
# Find the player who has maximum sixes and best strike rate
player_6s = cricket_data.loc[cricket_data['6s'].idxmax(), 'Player']

player_BatSR = cricket_data.loc[cricket_data['Bat_SR'].idxmax(), 'Player']

print("Player with maximum sixes:", player_6s)
print("Player with best strike rate:", player_BatSR)
```
